Amazon E-commerce Products & Reviews Dataset

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("C:\\Users\\anilr\\OneDrive\\Desktop\\Amazon E-commerce Products & Reviews Dataset\\Amazon_E-commerce_Products_clean_data.csv")
df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
print(df.shape)
print(df.head())
print(df.info())
print(df.describe())
```

```python
sns.histplot(data=df, x='popularity', bins=20)

plt.title('Distribution of Song Popularity')
plt.xlabel('Popularity')
plt.ylabel('Number of Songs')
plt.show()
```

```python
sns.histplot(data=df, x='danceability', bins=20)

plt.title('Distribution of Danceability')
plt.show()
```

```python
sns.histplot(data=df, x='energy', bins=20)

plt.title('Distribution of Energy')
plt.show()
```

```python
top_artists = df['artists'].value_counts().head(10)

sns.barplot(
    x=top_artists.values,
    y=top_artists.index
)

plt.title('Top 10 Artists by Number of Songs')
plt.xlabel('Number of Songs')
plt.ylabel('Artist')
plt.show()
```

```python
genre_popularity = (
    df.groupby('track_genre')['popularity']
      .mean()
      .sort_values(ascending=False)
      .head(15)
)

sns.barplot(
    x=genre_popularity.values,
    y=genre_popularity.index
)

plt.title('Top 15 Genres by Average Popularity')
plt.xlabel('Average Popularity')
plt.ylabel('Genre')
plt.show()
```

```python
numeric_cols = [
    'popularity',
    'duration_ms',
    'danceability',
    'energy',
    'loudness',
    'speechiness',
    'acousticness',
    'instrumentalness',
    'liveness',
    'valence',
    'tempo'
]

corr = df[numeric_cols].corr()

plt.figure(figsize=(12, 8))
sns.heatmap(corr, annot=True, fmt='.2f')

plt.title('Correlation Heatmap')
plt.show()
```

```python
sns.scatterplot(
    data=df,
    x='danceability',
    y='popularity'
)

plt.title('Popularity vs Danceability')
plt.show()
```

```python
sns.scatterplot(
    data=df,
    x='energy',
    y='popularity'
)

plt.title('Popularity vs Energy')
plt.show()
```

```python
sns.scatterplot(
    data=df,
    x='valence',
    y='popularity'
)

plt.title('Popularity vs Valence')
plt.show()
```

```python
sns.countplot(data=df, x='explicit')

plt.title('Explicit vs Non-Explicit Songs')
plt.xlabel('Explicit')
plt.ylabel('Number of Songs')
plt.show()
```

```python
sns.barplot(
    data=df,
    x='explicit',
    y='popularity'
)

plt.title('Average Popularity: Explicit vs Non-Explicit')
plt.show()
```

```python
df['duration_min'] = df['duration_ms'] / 60000
sns.histplot(data=df, x='duration_min', bins=30)

plt.title('Distribution of Song Duration')
plt.xlabel('Duration (minutes)')
plt.show()
```

```python
genre_count = df['track_genre'].value_counts().head(15)

sns.barplot(
    x=genre_count.values,
    y=genre_count.index
)

plt.title('Top 15 Genres by Number of Songs')
plt.xlabel('Number of Songs')
plt.ylabel('Genre')
plt.show()
```

```python
sns.boxplot(data=df, x='popularity')

plt.title('Popularity Outliers')
plt.show()
```

```python
sns.boxplot(data=df, x='energy')

plt.title('Energy Outliers')
plt.show()
```
