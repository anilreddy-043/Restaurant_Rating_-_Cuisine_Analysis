Restaurant Rating & Cuisine Analysis cleanning

```python
# import the required libraries
import pandas as pd
import numpy as np

df = pd.read_csv("C:\\Users\\anilr\\OneDrive\\Desktop\\Restaurant_Rating_&_Cuisine_Analysis\\zomato_india_1000_restaurants.csv")
df.head(10)
```

```python
# convert the 'Rating' column to numeric values, replacing 'NEW' and '-' with NaN
df['Rating'] = pd.to_numeric(df['Rating'].replace(['NEW', '-'], np.nan))

# fillinf missing values in the 'Rating' column with the mean value of the column
df['Rating'] = df['Rating'].fillna(df['Rating'].mean())

# changing the isnull values in the 'Rating' column to 0
df['Rating'] = df['Rating'].fillna(0)

# findt the null values in the 'Rating' column
df['Rating'].isnull().sum()
```

```python
# convert the "Reviews" column to numeric values, replacing '-' with NaN
df['Reviews'] = pd.to_numeric(df['Reviews'].replace('-', np.nan))

# fillinf missing values in the 'Reviews' column with the mean value of the column
df['Reviews'] = df['Reviews'].fillna(df['Reviews'].mean())

# changing the isnull values in the 'Reviews' column to 0
df['Reviews'] = df['Reviews'].fillna(0) 

# findt the null values in the 'Reviews' column
df['Reviews'].isnull().sum()
```

```python
df.info()
```

```python
# filling the missing values in the Cuisine column with the mode value of the column
df['Cuisine'] = df['Cuisine'].fillna(df['Cuisine'].mode()[0])

# changing the isnull values in the 'Cuisine' column to 'Not Specified'
df['Cuisine'] = df['Cuisine'].fillna('Not Specified')   

# findt the null values in the 'Cuisine' column
df['Cuisine'].isnull().sum()
```

```python
df.isna().sum()
```

```python
df.dtypes
```

```python
# Rating column in between 0 to 5
df['Rating'] = df['Rating'].apply(lambda x: 0 if x < 0 else (5 if x > 5 else x))    
```

```python
# fing the outliers in the 'Rating' column using the IQR method
Q1 = df['Rating'].quantile(0.25)
Q3 = df['Rating'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
#print the lower and upper bounds
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
# find the outliers in the 'Reviews' column
Q1_reviews = df['Reviews'].quantile(0.25)
Q3_reviews = df['Reviews'].quantile(0.75)
IQR_reviews = Q3_reviews - Q1_reviews
lower_bound_reviews = Q1_reviews - 1.5 * IQR_reviews
upper_bound_reviews = Q3_reviews + 1.5 * IQR_reviews
#print the lower and upper bounds for 'Reviews' column
print("Lower Bound for Reviews:", lower_bound_reviews)
print("Upper Bound for Reviews:", upper_bound_reviews)
```
